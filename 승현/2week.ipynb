```python
import pandas as pd

# 예제 데이터프레임 생성
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, None, 30],
    'Score': [90, 85, None]
}
df = pd.DataFrame(data)

# 결측치 여부 확인
print("결측치 여부 확인:")
print(df.isnull())
```

## 이상치 확인

```python
# 데이터의 이상치를 확인할때 최소/최대값을 확인한다.

import pandas as pd

# 예제 데이터프레임 생성
data = {'Height': [150, 160, 170, 180, 350]}
df = pd.DataFrame(data)

# 데이터 출력
print("데이터:\n", df)

# 최대값과 최소값 확인
print("\nHeight의 최대값:", df['Height'].max())
print("Height의 최소값:", df['Height'].min())
```

```python
# IQR을 이용한 이상치 확인
Q1 = df['Height'].quantile(0.25)
Q3 = df['Height'].quantile(0.75)
IQR = Q3 - Q1 # IQR 계산

# 이상치 기준 계산
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# 이상치 확인
outliers = df[(df['Height'] < lower_bound) | (df['Height'] > upper_bound)]
print("IQR 기준으로 찾은 이상치:\n", outliers)
```

```python
# 이상치 제거 (그냥 예외시키고 조회한 꼴이긴 함)
cleaned_df = df[(df['Height'] > lower_bound) & (df['Height'] < upper_bound)]
cleaned_df
```

```python
import pandas as pd

# 데이터 생성
data = {'Age': [15, 22, 35, 50, 72]}
df = pd.DataFrame(data)

# 나이를 구간화 (10대, 20대, 30대, 40대 이상)
bins = [0, 20, 40, 60, 80] # 구간 기준
labels = ['10대', '20대', '30대', '40대 이상'] # 구간별로 이름 지정 리스
df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)


# 결과 출력
print(df)
```

```python
# 나이를 자동으로 3개의 구간으로 나누기
df['Age_Quantile'] = pd.qcut(df['Age'], q=3, labels=['하', '중', '상'])

# 결과 출력
print(df)
```

```python
from sklearn.preprocessing import MinMaxScaler #도구를 사들임
import pandas as pd

# 데이터 생성
data = {'Height': [150, 160, 170, 180, 190]}
df = pd.DataFrame(data)

# MinMaxScaler 적용
scaler = MinMaxScaler() # 도구를 지정
df['Height_Normalized'] = scaler.fit_transform(df[['Height']]) #도구를 씀

# 결과 출력
print(df)
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# 데이터 생성
data = {'Age': [15, 22, 35, 50, 72], 'Weight': [50, 60, 70, 80, 90]}
df = pd.DataFrame(data)

# 구간화
bins = [0, 20, 40, 60, 80]
labels = ['10대', '20대', '30대', '40대 이상']
df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)

# 정규화
minmax_scaler = MinMaxScaler()
df['Weight_Normalized'] = minmax_scaler.fit_transform(df[['Weight']])

# 표준화 --> 이게 좀 더 좋음 
std_scaler = StandardScaler()
df['Weight_Standardized'] = std_scaler.fit_transform(df[['Weight']])
print(df)
```

```python
import pandas as pd

# 예제 데이터프레임 생성
df = pd.DataFrame({
    'name': [' 김철수 ', '이영희', '박민수', '정수진', '이서준'],
    'email': ['kim@naver.com', 'lee@gmail.com', 'park@daum.net', 'jung@korea.kr', 'lee@a.kr'],
    'gender_eng': ['Male', 'Female', 'Male', 'Female', 'Male']
})

# (1) 문자열 길이 계산
df['name_length'] = df['name'].str.len()

# (2) 대문자/소문자 변환
df['gender_upper'] = df['gender_eng'].str.upper()
df['gender_lower'] = df['gender_eng'].str.lower()

# (3) 특정 문자 포함 여부 확인 (이름에 '수'가 포함됐는가?)
df['has_수'] = df['name'].str.contains('수')

# (4) 문자열 바꾸기 (영문 → 한글)
df['gender_kr'] = df['gender_eng'].str.replace('Male', '남성').str.replace('Female', '여성')

# (5) 문자열 자르기 (이름 첫 글자)
df['first_letter'] = df['name'].str.slice(0, 1)

# (6) 문자열 나누기: 이메일에서 도메인 추출
df['email_domain'] = df['email'].str.split('@').str[1]

# (7) 공백 제거
df['name_stripped'] = df['name'].str.strip()

# (8) 문자열 연결: 이름 + 도메인 합치기
df['name_email_concat'] = df['name_stripped'] + '_' + df['email_domain']

print(df[['name', 'name_length', 'name_stripped', 'first_letter', 
          'gender_eng', 'gender_upper', 'gender_lower', 'gender_kr', 
          'email', 'email_domain', 'name_email_concat']])
```
