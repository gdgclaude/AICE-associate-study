```python
import numpy as np
```

```python
array = np.array([1,2,3,4,5]) # 1차원 배열 만들기
print(array)
```

```python
print("배열의 자료형:", array.dtype) # 배열의 자료형 반환
```

```python
matrix = np.array([[1,2,3], [4,5,6]]) # 2행 3열의 배열 생성 (메트릭스)
print(matrix)
```

```python
# 그 외 matrix data EDA 하는법
print("배열의 크기:", matrix.size)
print("배열의 형태:", matrix.shape)
print("배열의 차원", matrix.ndim)
```

```python
# 배열의 연산 (이 연산 기능을 매우 빠르게함)
print("배열의 합:", np.sum(matrix))
print("배열의 평균:", np.mean(matrix))
print("배열의 최대값:", np.max(matrix))
print("배열의 최소값:", np.min(matrix))
```

```python
# 배열 형태 바꾸기 reshape
reshaped_matrix = matrix.reshape(3,2)
print(reshaped_matrix)
```

```python
# 배열의 전치 T
transposed = matrix.T
print("전치된 배열:", transposed)   # (2,3) --> (3,2) 행과 열이 바뀌는걸 전치라 함
```

```python
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
print(type(fruits)) # (class ’list')
```

```python
print(fruits[0]) # apple (첫 번째 요소)
print(fruits[-1]) # cherry (마지막 요소)
```

```python
fruits.append("orange") # 요소 추가
print(fruits) # ['apple’, banana', 'cherry, ‘orange]
fruits.remove("banana") # 특정 _오소 삭제
print(fruits) # [‘apple’, cherry‘, ‘orange]
```

```python
# 튜블 (tuple)
colors = ( "red" , "green" , "blue" )
print(colors[1]) # green
# colors[1] = "yellow“ # 오류 발생 (튜플은 변경 불가)
```

```python
person = {"name": "Alice", "age" : 25, "city": "Seoul"}
print(person["name"]) # Alice
person["age"] = 26 # 값 변경
print(person) # {'name'Alice', 'age'： 26, city： 'Seoul'}
```

```python
# 집합(set)
numbers = {1, 2, 3, 4, 4, 5}
print(numbers) # //, 2, 3. 4, 5} (중복 제거)
```
