```python
import pandas as pd

## 1. 행 삭제 
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age' : [25, 30, 35, 40],
        'Score' : [90, 85, 75, 60] }
df = pd.DataFrame(data)

df_dropped = df.drop(index=2) #2행 삭제 
print(df_dropped) #표를 출력함 
```

```python
import pandas as pd

## 2. 열 삭제
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age' : [25, 30, 35, 40],
        'Score' : [90, 85, 75, 60] }
df = pd.DataFrame(data)

# drop(columns=['피쳐']) -> 특정 피쳐 칼럼 삭제
df_dropped = df.drop(columns=['Score']) #Score 열 삭제 
print(df_dropped) #표를 출력함 
```

```python
import pandas as pd

## 3. 결측치 행 삭제
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age' : [25, None, 35, 40],
        'Score' : [90, 85, None, 60] }
df_with_na = pd.DataFrame(data)

#dropna() -> 결측치 있는 행 삭제
df_dropped = df_with_na.dropna() #결측치가 있는 행 삭제
print(df_dropped) #표를 출력함 
```

```python
import pandas as pd

data = {'Name': ['Alice', 'Bob', 'Alice', 'David'],
        'Age' : [25, 30, 25, 40],
        'Score' : [90, 85, 90, 60] }
df_with_duplicates = pd.DataFrame(data)

#drop_duplicates() -> 중복된 행 삭제
df_unique = df_with_duplicates.drop_duplicates() 

print(df_unique) #표를 출력함 
```

## 데이터 전처리 함수 
- 1. 수치형 데이터 파생 함수
- 2. 문자형 데이터 파생 함수

```python
import pandas as pd 

df = pd.read_csv(file_path)

df['heart_age_ratio'] = (df['thalachh'] / df['age']).round(2) # 소수점 2번 째 자리에서 반올림
df['chol_cumsum'] = df['chol'].cumsum() # cumsum() - 누적합
df['age_group'] = (df['age'] // 10) * 10
df['thalach_rank'] = df['thalachh'].rank()
df['chol_diff'] = df['chol'].diff() # 변화량 출력

print(df)
```

```python
import pandas as pd 

df = pd.DataFrame({
  'name' : ['김철수', '이영희', '박민수', '정수진', '이서준'],
  'email' : ['kim@naver.com', 'lee@gmail.com', 'park@daum.net', 'jung@hanmail.net', 'seo@korea.kr'],
  'gender_eng': ["Male", 'Female', 'Male', 'Female', 'Male']
})

# 1. 문자열 길이 계산
df['name_length'] = df['name'].str.len()

# 2. 대문자/소문자 변환
df['gender_upper'] = df['gender_eng'].str.upper()
df['gender_lower'] = df['gender_eng'].str.lower()

# 3. 특정 문자 포함 여부 확인 (이름에 '수'가 포함됐는가?)
df['has_수'] = df['name'].str.contains('수')

# 4. 문자열 바꾸기
df['gender_kr'] = df['gender_eng'].str.replace('Male', '남성').str.replace('Female', '여성')

# 5. 문자열 자르기
df['first_letter'] = df['name'].str.slice(0,1)

# 6. 문자열 나누기
df['email_domain'] = df['email'].str.split('@').str[1]

# 7. 공백 제거
df['name_stripped'] = df['name'].str.strip()

# 8. 문자열 연결: 이름 + 도메인 합치기
df['name_email_concat'] = df['name_stripped'] + '_' + df['email_domain']

print(df[['name',
    'name_length',
    'gender_upper',
    'gender_lower',
    'has_수',
    'gender_kr',
    'first_letter',
    'email_domain',
    'name_stripped',
    'name_email_concat'
]])
```

```python
import pandas as pd

# (1) 예제 데이터 생성
df = pd.DataFrame({
    '가입일': pd.to_datetime(['2022-01-15', '2023-03-10', '2023-12-01', '2024-05-03', '2024-10-20'])
})
# (2) 기준일 설정
df['today'] = pd.to_datetime('2025-05-05')

# (3) 날짜 파생변수 생성
df['가입연도'] = df['가입일'].dt.year              # 연도
df['가입월'] = df['가입일'].dt.month              # 월
df['가입일자'] = df['가입일'].dt.day              # 일
df['가입요일'] = df['가입일'].dt.day_name()       # 요일명
df['가입요일번호'] = df['가입일'].dt.weekday       # 0=월 ~ 6=일
df['가입_분기'] = df['가입일'].dt.quarter          # 1~4분기
df['가입_주차'] = df['가입일'].dt.isocalendar().week  # ISO 기준 주차
df['월_시작일인가'] = df['가입일'].dt.is_month_start # 월 첫날?
df['월_마지막일인가'] = df['가입일'].dt.is_month_end # 월 마지막일?
df['가입후_경과일수'] = (df['today'] - df['가입일']).dt.days  # 경과 일수

# (4) 주말 여부 판단
df['주말가입여부'] = df['가입요일번호'].isin([5, 6])  # 토/일 = True

print(df)
```
