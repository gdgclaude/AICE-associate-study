## Matplotlib 라이브러리 사용하기

```python
import matplotlib.pyplot as plt
from matplotlib import rc
import warnings # 경고 메시지를 제거할 때 사용

# 프로그램 실행 중 경고 메시지 제거 
warnings.filterwarnings("ignore")

# 한글 글꼴 설정
rc('font', family='Malgun Gothic')


x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.plot(x, y, label ='한글 데이터', color = 'blue', marker='o') # 그래프 그리기
plt.title("한글과 경고 제거 테스트") # 그래프의 제목 설정
plt.xlabel('X축 이름') 
plt.ylabel('Y축 이름')
plt.legend() # label로 설정한 내용을 그래프 옆에 범례로 표시
plt.grid(True) # 그래프에 격자 추가
plt.show()
```

## Seaborn
> matplotlib의 상위 레이어로, 더 간단하고 직관적인 방식으로 시각화를 제공한다. 데이터 프레임과의 호환성이 뛰어나 데이터 분석과 정에서 널리 사용됨

```python
import seaborn as sns
import pandas as pd

data = pd.DataFrame({
  'Category': ['A', 'B', 'C', 'D'],
  'Values' : [5, 6, 3, 8]
})

# 막대그래프
sns.barplot(x='Category', y = 'Values', data=data) # 막대의 이름을 지정하는 열, 각 카테고리에 해당한는 값으로 막대 그래프의 높이를 결정
plt.title("Bar Plot Example")
plt.show()
```

## plotly 라이브러리 활용하기
> 정적 그래프를 넘어 동적이고 직관적인 시각화를 구현할 수 있는 도구이다.

```python
import plotly.express as px

# 샘플 데이터
data = px.data.gapminder()

# 대화형 산점도
fig =px.scatter(
  data,
  x='gdpPercap',
  y='lifeExp',
  size='pop',  # 점의 크기를 나타낼 열을 선택 
  color='continent', 
  log_x=True, # x축의 값을 로그 스케일로 변환
  title="Interactive Scatter Plot"
)
fig.update_layout(width=800, height=500) 
fig.show()
```

```python
import pandas as pd

# 샘플 데이터프레임 생성
data = {'Year': [2020, 2021, 2022], 'Sales': [250, 300, 400]}
df = pd.DataFrame(data)


df.plot(x='Year', y='Sales', kind='line', title="Sales Over Years") # x축 지정, y축 지정, kind 그래프의 종류 지정 = 선그래프 
plt.show()
```

```python
# 2. 선 그래프
import pandas as pd 
import matplotlib.pyplot as plt

df = pd.read_csv("../datasets/heart.csv")

age_chol = df.groupby("age")['chol'].mean().reset_index()


# 그래프 그리기 
plt.plot(age_chol['age'], age_chol['chol'])
plt.xlabel('나이')
plt.ylabel('평균 콜레스테롤 수치')
plt.title('나이 별 콜레스테롤 변화')
plt.grid(True)
plt.show()
```

```python
plt.bar(df['sex'], df['chol'])
plt.xlabel('성별')
plt.ylabel('콜레스테롤 수치')
plt.title('성별에 따른 콜레스테롤 수치')
plt.show()
```

```python
plt.scatter(df['age'], df['thalachh'])
plt.xlabel('나이')
plt.ylabel('최대 심박수')
plt.title('나이와 심박수의 관계')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 데이터 불러오기
data = pd.read_csv("../datasets/heart.csv")

# 히스토그램 생성
plt.hist(
    data['age'],          # 나이 데이터를 시각화
    bins=10,              # 구간 수
    color='skyblue',      # 막대 색상
    alpha=0.7,            # 투명도
    edgecolor='black'     # 테두리 색상
)
plt.title("Age Distribution in Heart Dataset")  # 그래프 제목
plt.xlabel("Age")                               # X축 이름
plt.ylabel("Frequency")                         # Y축 이름
plt.grid(True)                                  # 격자 표시
plt.show()
```

```python
import pandas as pd 
import matplotlib.pyplot as plt

data = pd.read_csv('../datasets/heart.csv')

plt.boxplot(
  data['age'],
  vert=True, # 그래프를 세로로 표시
  patch_artist=True,
  boxprops=dict(facecolor='skyblue', color='black'),
  medianprops=dict(color='red', linewidth=2),
  whiskerprops=dict(color='black', linewidth=1.5)
)


plt.title("Age Disribution in Heart Dataset")
plt.ylabel("Age")
plt.grid(True)
plt.show()
```

```python
import pandas as pd 
import matplotlib.pyplot as plt

data = pd.read_csv('../datasets/heart.csv')

data = data.sort_values(by='age')

plt.plot(
  data['age'],
  data['chol'],
  color='blue',
  linewidth=2,
  linestyle='-',
  marker='o',
  markersize=5,
  label='Colesterol'
)

plt.title("Age vs Colesterol")
plt.xlabel("Age")
plt.ylabel("Colesterol")
plt.legend()
plt.grid(True)
plt.show()
```
